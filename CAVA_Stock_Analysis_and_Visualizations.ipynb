# CAVA Stock Analysis (June 15–August 15, 2026)
This notebook downloads CAVA daily prices, calculates daily returns, and creates three charts. August 15, 2026 was a Saturday, so August 14 is the last trading session in this window.

```python
%pip -q install yfinance pandas matplotlib
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

ticker = "CAVA"
prices = yf.download(
    ticker,
    start="2026-06-15",
    end="2026-08-16",  # end date is exclusive
    auto_adjust=False,
    progress=False
)

if prices.empty:
    raise RuntimeError("No price data was returned. Check your internet connection and try again.")

# yfinance may return MultiIndex columns; flatten them for easier use.
if isinstance(prices.columns, pd.MultiIndex):
    prices.columns = prices.columns.get_level_values(0)

prices = prices.sort_index()
prices["Daily_Return_Pct"] = prices["Close"].pct_change() * 100
prices["Cumulative_Return_Pct"] = (prices["Close"] / prices["Close"].iloc[0] - 1) * 100

display(prices.head())
display(prices.tail())

period_return = (prices["Close"].iloc[-1] / prices["Close"].iloc[0] - 1) * 100
print(f"First close: ${prices['Close'].iloc[0]:.2f}")
print(f"Last close:  ${prices['Close'].iloc[-1]:.2f}")
print(f"Period return: {period_return:.2f}%")
```

```python
# Chart 1: Closing price
plt.figure(figsize=(12, 5))
plt.plot(prices.index, prices["Close"], marker="o")
plt.title("CAVA Daily Closing Price (June 15–August 15, 2026)")
plt.xlabel("Date")
plt.ylabel("Closing Price (USD)")
plt.grid(True, alpha=0.3)
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Daily close-to-close returns
daily_returns = prices["Daily_Return_Pct"].dropna()
plt.figure(figsize=(12, 5))
plt.bar(daily_returns.index, daily_returns.values)
plt.axhline(0, linewidth=1)
plt.title("CAVA Daily Stock Returns")
plt.xlabel("Date")
plt.ylabel("Daily Return (%)")
plt.grid(True, axis="y", alpha=0.3)
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Cumulative return from the first trading day
plt.figure(figsize=(12, 5))
plt.plot(prices.index, prices["Cumulative_Return_Pct"], linewidth=2)
plt.axhline(0, linewidth=1)
plt.title("CAVA Cumulative Return Since June 15, 2026")
plt.xlabel("Date")
plt.ylabel("Cumulative Return (%)")
plt.grid(True, alpha=0.3)
plt.xticks(rotation=35)
plt.tight_layout()
plt.show()
```

```python
# Save a CSV file that can be shared with your group
csv_file = "CAVA_daily_data_June15_Aug15_2026.csv"
prices.to_csv(csv_file)
print("Saved:", csv_file)
```
